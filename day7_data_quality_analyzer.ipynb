```python
import pandas as pd

def check_data_quality(df):
    results = {
        "Check": [
            "Total Records",
            "Missing Values",
            "Duplicate Rows",
            "Invalid Quantities",
            "Invalid Prices"
        ],
        "Result": [
            len(df),
            df.isnull().sum().sum(),
            df.duplicated().sum(),
            (df["quantity"] <= 0).sum(),
            (df["price"] <= 0).sum()
        ]
    }

    report = pd.DataFrame(results)

    report["Status"] = [
        "INFO",
        "PASS" if results["Result"][1] == 0 else "FAIL",
        "PASS" if results["Result"][2] == 0 else "FAIL",
        "PASS" if results["Result"][3] == 0 else "FAIL",
        "PASS" if results["Result"][4] == 0 else "FAIL"
    ]

    report.to_csv("day7_data_quality_report.csv", index=False)

    return report


df = pd.read_csv("sales_cleaned_day3.csv")

quality_report = check_data_quality(df)

print("Report saved successfully!")
quality_report
```

```python
import pandas as pd

def check_data_quality(df):

    missing = df.isnull().sum().sum()
    duplicates = df.duplicated().sum()
    invalid_quantity = (df["quantity"] <= 0).sum()
    invalid_price = (df["price"] <= 0).sum()

    report = pd.DataFrame({
        "Check": [
            "Total Records",
            "Missing Values",
            "Duplicate Rows",
            "Invalid Quantities",
            "Invalid Prices"
        ],
        "Result": [
            len(df),
            missing,
            duplicates,
            invalid_quantity,
            invalid_price
        ],
        "Status": [
            "INFO",
            "PASS" if missing == 0 else "FAIL",
            "PASS" if duplicates == 0 else "FAIL",
            "PASS" if invalid_quantity == 0 else "FAIL",
            "PASS" if invalid_price == 0 else "FAIL"
        ]
    })

    report.to_csv("day7_data_quality_report.csv", index=False)

    return report
```
