```python
import pandas as pd

df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
# Count quality checks
total_checks = 4

passed_checks = 0

if df.isnull().sum().sum() == 0:
    passed_checks += 1

if df.duplicated().sum() == 0:
    passed_checks += 1

if (df["quantity"] > 0).all():
    passed_checks += 1

if (df["price"] > 0).all():
    passed_checks += 1

# Calculate percentage
quality_score = (passed_checks / total_checks) * 100

print("Total checks:", total_checks)
print("Passed checks:", passed_checks)
print("Data Quality Score:", quality_score, "%")
```

```python
# Load the second dataset
df2 = pd.read_csv("sample_sales_2.csv")

# Check 4 quality rules
checks = {
    "Missing Values": df2.isnull().sum().sum() == 0,
    "Duplicate Rows": df2.duplicated().sum() == 0,
    "Valid Quantities": (df2["quantity"] > 0).all(),
    "Valid Prices": (df2["price"] > 0).all()
}

passed_checks = sum(checks.values())
total_checks = len(checks)

quality_score = (passed_checks / total_checks) * 100

print("QUALITY CHECK RESULTS")
print("---------------------")

for check, passed in checks.items():
    print(check, ":", "PASS" if passed else "FAIL")

print("\nPassed checks:", passed_checks)
print("Total checks:", total_checks)
print("Data Quality Score:", quality_score, "%")
```

```python
import pandas as pd

def calculate_quality_score(data):
    checks = {
        "Missing Values": data.isnull().sum().sum() == 0,
        "Duplicate Rows": data.duplicated().sum() == 0,
        "Valid Quantities": (data["quantity"] > 0).all(),
        "Valid Prices": (data["price"] > 0).all()
    }

    passed = sum(checks.values())
    total = len(checks)
    score = (passed / total) * 100

    return passed, total, score


files = [
    "sales_cleaned_day3.csv",
    "sample_sales_2.csv"
]

results = []

for file in files:
    data = pd.read_csv(file)
    passed, total, score = calculate_quality_score(data)

    results.append({
        "File Name": file,
        "Passed Checks": passed,
        "Total Checks": total,
        "Quality Score (%)": score
    })

comparison_report = pd.DataFrame(results)

comparison_report.to_csv(
    "day9_quality_score_comparison.csv",
    index=False
)

print(comparison_report.to_string(index=False))
print("\nComparison report saved successfully!")
```
