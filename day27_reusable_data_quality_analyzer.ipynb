```python

import pandas as pd
import os

print("Day 27: Reusable Data Quality Analyzer")
print("Project folder:", os.getcwd())
```

```python
df = pd.read_csv("day26_cleaned_sales.csv")

print("Dataset loaded successfully!")
print("Total records:", len(df))
print("Total columns:", len(df.columns))

display(df)
```

```python

def analyze_data_quality(df):
    report = pd.DataFrame({
        "Check": [
            "Total Records",
            "Missing Values",
            "Duplicate Rows",
            "Invalid Quantities",
            "Invalid Prices"
        ],
        "Result": [
            len(df),
            int(df.isnull().sum().sum()),
            int(df.duplicated().sum()),
            int((df["quantity"] <= 0).sum()),
            int((df["price"] <= 0).sum())
        ]
    })

    report["Status"] = report.apply(
        lambda row: (
            "INFO" if row["Check"] == "Total Records"
            else "PASS" if row["Result"] == 0
            else "FAIL"
        ),
        axis=1
    )

    return report

print("Reusable function created successfully!")
```

```python
quality_report = analyze_data_quality(df)

print("Data Quality Analysis Completed!")
display(quality_report)
```

```python
quality_report.to_csv(
    "day27_reusable_quality_report.csv",
    index=False
)

print("Quality report saved successfully!")
print("File: day27_reusable_quality_report.csv")
```

```python
df2 = pd.read_csv("sample_sales_2.csv")

report1 = analyze_data_quality(df)
report2 = analyze_data_quality(df2)

print("Dataset 1: Cleaned Sales Data")
display(report1)

print("Dataset 2: Sample Sales Data")
display(report2)
```

```python
report1["Dataset"] = "day26_cleaned_sales.csv"
report2["Dataset"] = "sample_sales_2.csv"

combined_report = pd.concat(
    [report1, report2],
    ignore_index=True
)

combined_report = combined_report[
    ["Dataset", "Check", "Result", "Status"]
]

print("Combined quality report created!")
display(combined_report)
```

```python
combined_report.to_csv(
    "day27_combined_quality_report.csv",
    index=False
)

print("Combined report saved successfully!")
print("File: day27_combined_quality_report.csv")
print("Total report rows:", len(combined_report))
```
