```python
import pandas as pd
import os

print("Day 11: Multiple CSV Data Quality Analysis")
print("Current folder:", os.getcwd())
```

```python
new_data = {
    "order_id": [201, 202, 203, 204],
    "product": ["Phone", "Tablet", None, "Laptop"],
    "quantity": [2, 1, 3, 0],
    "price": [15000, 25000, 500, 40000],
    "city": ["Jaipur", "Delhi", "Mumbai", "Jaipur"]
}

new_df = pd.DataFrame(new_data)

new_df.to_csv("sample_sales_2.csv", index=False)

print("Second CSV file created successfully!")
print(new_df)
```

```python
files = [
    "sales_cleaned_day3.csv",
    "sample_sales_2.csv"
]

for file in files:
    df = pd.read_csv(file)

    print("\nFile Name:", file)
    print("Total Rows:", len(df))
    print(df.head())
```

```python
from data_quality_analyzer import check_data_quality

all_reports = []

for file in files:
    df = pd.read_csv(file)

    report = check_data_quality(df)
    report.insert(0, "File Name", file)

    all_reports.append(report)

combined_report = pd.concat(all_reports, ignore_index=True)

print(combined_report.to_string(index=False))
```

```python
combined_report.to_csv(
    "day11_combined_quality_report.csv",
    index=False
)

print("Combined report saved successfully!")
print("Filename: day11_combined_quality_report.csv")
```
