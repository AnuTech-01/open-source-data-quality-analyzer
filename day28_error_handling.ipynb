```python
import pandas as pd
import os

print("Day 28: Error Handling")
print("Project folder:", os.getcwd())
```

```python
file_name = "day26_cleaned_sales.csv"

try:
    df = pd.read_csv(file_name)
    print("✅ File loaded successfully!")
    print("Total records:", len(df))
    display(df.head())

except FileNotFoundError:
    print(f"❌ File not found: {file_name}")

except pd.errors.ParserError:
    print("❌ CSV file format mein problem hai.")

except Exception as e:
    print("❌ Unexpected error:", e)
```

```python
required_columns = ["order_id", "product", "quantity", "price", "city"]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    print("❌ Required columns missing:", missing_columns)
else:
    print("✅ All required columns are available!")

print("Available columns:", list(df.columns))
```

```python
for column in ["quantity", "price"]:
    converted = pd.to_numeric(df[column], errors="coerce")
    invalid_count = int(
        (converted.isna() & df[column].notna()).sum()
    )

    if invalid_count > 0:
        print(f"❌ {column}: {invalid_count} invalid value(s) found")
    else:
        print(f"✅ {column}: All non-missing values are numeric")
```

```python
error_checks = []

# Check required columns
error_checks.append({
    "Check": "Required Columns",
    "Status": "PASS" if not missing_columns else "FAIL",
    "Details": (
        "All required columns available"
        if not missing_columns
        else f"Missing columns: {missing_columns}"
    )
})

# Check numeric data types
for column in ["quantity", "price"]:
    converted = pd.to_numeric(df[column], errors="coerce")
    invalid_count = int(
        (converted.isna() & df[column].notna()).sum()
    )

    error_checks.append({
        "Check": f"{column} Data Type",
        "Status": "PASS" if invalid_count == 0 else "FAIL",
        "Details": f"{invalid_count} invalid value(s)"
    })

error_report = pd.DataFrame(error_checks)

display(error_report)
```

```python
error_report.to_csv(
    "day28_error_handling_report.csv",
    index=False
)

print("✅ Error handling report saved successfully!")
print("File: day28_error_handling_report.csv")
print("Total checks:", len(error_report))
```
