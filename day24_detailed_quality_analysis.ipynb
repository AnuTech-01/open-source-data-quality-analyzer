```python
import pandas as pd
import os

print("Day 24: Detailed Data Quality Analysis")
print("Project folder:", os.getcwd())
```

```python
df = pd.read_csv("sample_sales_2.csv")

print("Dataset loaded successfully!")
print("Total rows:", len(df))

display(df)
```

```python
missing_by_column = df.isnull().sum()
duplicate_rows = df.duplicated().sum()
invalid_quantities = (df["quantity"] <= 0).sum()
invalid_prices = (df["price"] <= 0).sum()

print("Missing Values by Column:")
display(missing_by_column)

print("Duplicate Rows:", int(duplicate_rows))
print("Invalid Quantities:", int(invalid_quantities))
print("Invalid Prices:", int(invalid_prices))
```

```python

detailed_report = pd.DataFrame({
    "Quality Issue": [
        "Missing Values",
        "Duplicate Rows",
        "Invalid Quantities",
        "Invalid Prices"
    ],
    "Count": [
        int(df.isnull().sum().sum()),
        int(duplicate_rows),
        int(invalid_quantities),
        int(invalid_prices)
    ]
})

detailed_report["Status"] = detailed_report["Count"].apply(
    lambda count: "PASS" if count == 0 else "FAIL"
)

display(detailed_report)
```

```python
detailed_report.to_csv(
    "day24_detailed_quality_report.csv",
    index=False
)

print("Detailed quality report saved successfully!")
print("File: day24_detailed_quality_report.csv")
```

```python
file_name = "day24_detailed_quality_report.csv"

if os.path.exists(file_name):
    saved_report = pd.read_csv(file_name)

    print("Day 24 Completed Successfully!")
    print("Report file:", file_name)
    print("Total quality checks:", len(saved_report))
    display(saved_report)
else:
    print("Report not found. Please run the save cell again.")
```
