```python
import pandas as pd
import os

print("Day 13: Data Quality Summary")
print("Current folder:", os.getcwd())
```

```python
report_df = pd.read_csv("day12_automatic_quality_report.csv")

print("Report loaded successfully!")
print("Total rows:", len(report_df))
print(report_df.head())
```

```python
pass_count = (report_df["Status"] == "PASS").sum()
fail_count = (report_df["Status"] == "FAIL").sum()
info_count = (report_df["Status"] == "INFO").sum()

print("DATA QUALITY SUMMARY")
print("--------------------")
print("Total checks:", len(report_df))
print("PASS checks:", pass_count)
print("FAIL checks:", fail_count)
print("INFO checks:", info_count)
```

```python
summary_data = {
    "Metric": [
        "Total Checks",
        "PASS Checks",
        "FAIL Checks",
        "INFO Checks"
    ],
    "Count": [
        len(report_df),
        pass_count,
        fail_count,
        info_count
    ]
}

summary_df = pd.DataFrame(summary_data)

summary_df.to_csv("day13_quality_summary.csv", index=False)

print("Summary saved successfully!")
print(summary_df)
```
