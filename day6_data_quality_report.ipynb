```python
import pandas as pd

df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset loaded successfully!")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

df.head()
```

```python
# 1. Check missing values
missing_values = df.isnull().sum()
print("Missing values:")
print(missing_values)

# 2. Check duplicate rows
duplicate_count = df.duplicated().sum()
print("\nDuplicate rows:", duplicate_count)

# 3. Check invalid quantities
invalid_quantity = (df["quantity"] <= 0).sum()
print("Invalid quantities:", invalid_quantity)

# 4. Check invalid prices
invalid_price = (df["price"] <= 0).sum()
print("Invalid prices:", invalid_price)
```

```python
import pandas as pd

quality_report = pd.DataFrame({
    "Check": [
        "Total Records",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Invalid Quantities",
        "Invalid Prices"
    ],
    "Result": [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        (df["quantity"] <= 0).sum(),
        (df["price"] <= 0).sum()
    ]
})

quality_report["Status"] = [
    "PASS" if quality_report.loc[i, "Result"] >= 0 else "FAIL"
    for i in range(len(quality_report))
]

quality_report
```

```python
quality_report["Status"] = [
    "INFO",  # Total Records
    "INFO",  # Total Columns
    "PASS" if df.isnull().sum().sum() == 0 else "FAIL",
    "PASS" if df.duplicated().sum() == 0 else "FAIL",
    "PASS" if (df["quantity"] <= 0).sum() == 0 else "FAIL",
    "PASS" if (df["price"] <= 0).sum() == 0 else "FAIL"
]

quality_report
```

```python
quality_report.to_csv(
    "day6_data_quality_report.csv",
    index=False
)

print("Data Quality Report saved successfully!")
```
