```python
import pandas as pd
from data_quality_analyzer import check_data_quality

print("Day 18: Multi-Dataset Quality Analyzer")
print("Imports successful!")
```

```python
df1 = pd.read_csv("sales_cleaned_day3.csv")
df2 = pd.read_csv("sample_sales_2.csv")

print("Dataset 1: sales_cleaned_day3.csv")
print("Rows:", len(df1))

print("\nDataset 2: sample_sales_2.csv")
print("Rows:", len(df2))
```

```python
report1 = check_data_quality(df1)
report2 = check_data_quality(df2)

print("Quality Report - Dataset 1")
display(report1)

print("Quality Report - Dataset 2")
display(report2)
```

```python

report1["Dataset"] = "sales_cleaned_day3.csv"
report2["Dataset"] = "sample_sales_2.csv"

combined_report = pd.concat(
    [report1, report2],
    ignore_index=True
)

display(combined_report)
```

```python
combined_report.to_csv(
    "day18_multi_dataset_quality_report.csv",
    index=False
)

print("Day 18 report saved successfully!")
print("File: day18_multi_dataset_quality_report.csv")
```
