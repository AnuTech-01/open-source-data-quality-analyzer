```python
import pandas as pd

df = pd.read_csv("sales_without_duplicates.csv")

print("Dataset loaded successfully!")
print(df.head())

print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
df["product"] = df["product"].fillna("Unknown")
df["city"] = df["city"].fillna("Unknown")

print("Missing values after filling:")
print(df.isnull().sum())
```

```python
print("Quantity median:", df["quantity"].median())
```

```python
df["quantity"] = df["quantity"].fillna(df["quantity"].median())

print("Missing values after filling:")
print(df.isnull().sum())

print("\nUpdated quantity column:")
print(df["quantity"])
```

```python
df.to_csv("sales_cleaned_day3.csv", index=False)

print("Day 3 cleaned dataset saved successfully!")
```

```python
check_df = pd.read_csv("sales_cleaned_day3.csv")

print("Rows:", len(check_df))
print("Total missing values:", check_df.isnull().sum().sum())
print("Duplicate rows:", check_df.duplicated().sum())
```
