```python
import pandas as pd
import os

print("Day 25: Automatic Issue Detection")
print("Project folder:", os.getcwd())
```

```python
df = pd.read_csv("sample_sales_2.csv")

print("Dataset loaded successfully!")
print("Total records:", len(df))

display(df)
```

```python
issues = {
    "Missing Values": int(df.isnull().sum().sum()),
    "Duplicate Rows": int(df.duplicated().sum()),
    "Invalid Quantities": int((df["quantity"] <= 0).sum()),
    "Invalid Prices": int((df["price"] <= 0).sum())
}

print("Automatic Issue Detection Results:")

for issue, count in issues.items():
    if count > 0:
        print(f"{issue}: {count} issue(s) found")
    else:
        print(f"{issue}: No issues found")
```

```python

issue_report = pd.DataFrame({
    "Issue": list(issues.keys()),
    "Count": list(issues.values())
})

issue_report["Status"] = issue_report["Count"].apply(
    lambda count: "Issue Found" if count > 0 else "Clean"
)

issue_report["Recommended Action"] = issue_report["Issue"].map({
    "Missing Values": "Fill or remove missing values",
    "Duplicate Rows": "Review and remove duplicate records",
    "Invalid Quantities": "Check zero or negative quantities",
    "Invalid Prices": "Check zero or negative prices"
})

display(issue_report)
```

```python
issue_report.to_csv(
    "day25_automatic_issue_report.csv",
    index=False
)

print("Issue report saved successfully!")
print("File: day25_automatic_issue_report.csv")
```
