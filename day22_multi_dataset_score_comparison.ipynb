```python
import pandas as pd
import os

print("Day 22: Multi-Dataset Quality Score Comparison")
print("Project folder:", os.getcwd())
```

```python
df1 = pd.read_csv("sales_cleaned_day3.csv")
df2 = pd.read_csv("sample_sales_2.csv")

print("Dataset 1: sales_cleaned_day3.csv")
print("Rows:", len(df1))

print("\nDataset 2: sample_sales_2.csv")
print("Rows:", len(df2))
```

```python
def calculate_quality_score(df):
    missing = int(df.isnull().sum().sum())
    duplicates = int(df.duplicated().sum())
    invalid_quantity = int((df["quantity"] <= 0).sum())
    invalid_price = int((df["price"] <= 0).sum())

    passed_checks = sum([
        missing == 0,
        duplicates == 0,
        invalid_quantity == 0,
        invalid_price == 0
    ])

    return (passed_checks / 4) * 100


score1 = calculate_quality_score(df1)
score2 = calculate_quality_score(df2)

print("Clean Dataset Score:", score1, "%")
print("Second Dataset Score:", score2, "%")
```

```python
comparison_report = pd.DataFrame({
    "Dataset": [
        "sales_cleaned_day3.csv",
        "sample_sales_2.csv"
    ],
    "Quality Score (%)": [
        score1,
        score2
    ]
})

comparison_report["Category"] = comparison_report["Quality Score (%)"].apply(
    lambda score: "Excellent" if score >= 90
    else "Good" if score >= 70
    else "Average" if score >= 50
    else "Poor"
)

display(comparison_report)
```
