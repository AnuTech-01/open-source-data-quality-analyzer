```python
import pandas as pd

df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset loaded successfully!")
print(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
print("Invalid quantity values:")
print(df[df["quantity"] <= 0])

print("\nInvalid price values:")
print(df[df["price"] <= 0])
```

```python
invalid_quantity_count = (df["quantity"] <= 0).sum()
invalid_price_count = (df["price"] <= 0).sum()

print("DATA VALIDATION SUMMARY")
print("-----------------------")
print("Total records:", len(df))
print("Invalid quantities:", invalid_quantity_count)
print("Invalid prices:", invalid_price_count)

if invalid_quantity_count == 0 and invalid_price_count == 0:
    print("Validation status: Passed")
else:
    print("Validation status: Issues found")
```

```python
validation_report = pd.DataFrame({
    "check": [
        "Total records",
        "Invalid quantities",
        "Invalid prices"
    ],
    "result": [
        len(df),
        (df["quantity"] <= 0).sum(),
        (df["price"] <= 0).sum()
    ]
})

validation_report.to_csv("day4_validation_report.csv", index=False)

print(validation_report)
print("\nValidation report saved successfully!")
```
