```python
import pandas as pd

print("Day 15: Quality Score Comparison")
```

```python
df1 = pd.read_csv("sales_cleaned_day3.csv")
df2 = pd.read_csv("sample_sales_2.csv")

print("Dataset 1: sales_cleaned_day3.csv")
print("Total records:", len(df1))

print("\nDataset 2: sample_sales_2.csv")
print("Total records:", len(df2))
```

```python
def calculate_quality_score(df):
    missing = int(df.isnull().sum().sum())
    duplicates = int(df.duplicated().sum())
    invalid_quantity = int((df["quantity"] <= 0).sum())
    invalid_price = int((df["price"] <= 0).sum())

    passed_checks = sum([
        missing == 0,
        duplicates == 0,
        invalid_quantity == 0,
        invalid_price == 0
    ])

    return (passed_checks / 4) * 100


score1 = calculate_quality_score(df1)
score2 = calculate_quality_score(df2)

print("Dataset 1 Quality Score:", score1, "%")
print("Dataset 2 Quality Score:", score2, "%")
```

```python
comparison = pd.DataFrame({
    "File Name": [
        "sales_cleaned_day3.csv",
        "sample_sales_2.csv"
    ],
    "Quality Score (%)": [
        score1,
        score2
    ]
})

comparison.to_csv(
    "day15_quality_score_comparison.csv",
    index=False
)

print("Comparison report saved successfully!")
print(comparison)
```
