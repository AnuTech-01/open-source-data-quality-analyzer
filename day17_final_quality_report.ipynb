```python
import pandas as pd
import os

print("Day 17: Final Data Quality Report")
print("Project folder:", os.getcwd())
```

```python
comparison = pd.read_csv("day15_quality_score_comparison.csv")

print("Quality score report loaded successfully!")
display(comparison)
```

```python
final_report = comparison.copy()

final_report["Result"] = final_report["Quality Score (%)"].apply(
    lambda score: "Excellent" if score >= 90
    else "Good" if score >= 70
    else "Needs Improvement"
)

final_report["Report Date"] = pd.Timestamp.today().strftime("%Y-%m-%d")

display(final_report)
```

```python

final_report.to_csv("day17_final_quality_report.csv", index=False)

print("Final report saved successfully!")
print("File name: day17_final_quality_report.csv")
```
