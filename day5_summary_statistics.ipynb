```python
import pandas as pd

df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset shape:", df.shape)
print("\nNumerical summary:")
print(df.describe())
```

```python
print("Average price:", df["price"].mean())
print("Minimum price:", df["price"].min())
print("Maximum price:", df["price"].max())
print("Total quantity:", df["quantity"].sum())
```

```python
total_sales = (df["quantity"] * df["price"]).sum()
average_price = df["price"].mean()
highest_price = df["price"].max()
lowest_price = df["price"].min()

print("BUSINESS INSIGHTS")
print("-----------------")
print("Total sales value:", total_sales)
print("Average product price:", average_price)
print("Highest product price:", highest_price)
print("Lowest product price:", lowest_price)
```
