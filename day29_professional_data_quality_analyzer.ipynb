```python
import pandas as pd
import os

print("Day 29: Professional Data Quality Analyzer")
print("Project folder:", os.getcwd())
```

```python
def load_csv_file(file_name):
    try:
        data = pd.read_csv(file_name)
        print("✅ File loaded successfully:", file_name)
        print("Total records:", len(data))
        return data

    except FileNotFoundError:
        print("❌ File not found:", file_name)

    except pd.errors.ParserError:
        print("❌ CSV format error:", file_name)

    except Exception as e:
        print("❌ Unexpected error:", e)

    return None

print("CSV loading function created successfully!")
```

```python
df = load_csv_file("day26_cleaned_sales.csv")

if df is not None:
    display(df.head())
else:
    print("Please check the CSV file name and location.")
```

```python
def validate_columns(df):
    required_columns = [
        "order_id", "product", "quantity", "price", "city"
    ]

    missing_columns = [
        column for column in required_columns
        if column not in df.columns
    ]

    if missing_columns:
        print("❌ Missing columns:", missing_columns)
        return False

    print("✅ All required columns are available!")
    return True


if df is not None:
    columns_valid = validate_columns(df)
```

```python

def generate_quality_report(df):
    report = pd.DataFrame({
        "Check": [
            "Total Records",
            "Missing Values",
            "Duplicate Rows",
            "Invalid Quantities",
            "Invalid Prices"
        ],
        "Result": [
            len(df),
            int(df.isnull().sum().sum()),
            int(df.duplicated().sum()),
            int((pd.to_numeric(df["quantity"], errors="coerce") <= 0).sum()),
            int((pd.to_numeric(df["price"], errors="coerce") <= 0).sum())
        ]
    })

    report["Status"] = report.apply(
        lambda row: (
            "INFO" if row["Check"] == "Total Records"
            else "PASS" if row["Result"] == 0
            else "FAIL"
        ),
        axis=1
    )

    return report

print("Quality report function created successfully!")
```

```python
if df is not None:
    if validate_columns(df):
        quality_report = generate_quality_report(df)

        print("✅ Data Quality Analysis Completed!")
        display(quality_report)
    else:
        print("❌ Analysis stopped because required columns are missing.")
else:
    print("❌ Analysis stopped because the CSV file could not be loaded.")
```

```python
if "quality_report" in globals():
    quality_report.to_csv(
        "day29_professional_quality_report.csv",
        index=False
    )

    print("✅ Quality report saved successfully!")
    print("File: day29_professional_quality_report.csv")
else:
    print("❌ Quality report not available. Run Step 6 first.")
```
