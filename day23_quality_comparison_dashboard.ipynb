```python

import pandas as pd
import os

df1 = pd.read_csv("sales_cleaned_day3.csv")
df2 = pd.read_csv("sample_sales_2.csv")

def calculate_quality_score(df):
    missing = int(df.isnull().sum().sum())
    duplicates = int(df.duplicated().sum())
    invalid_quantity = int((df["quantity"] <= 0).sum())
    invalid_price = int((df["price"] <= 0).sum())

    passed = sum([
        missing == 0,
        duplicates == 0,
        invalid_quantity == 0,
        invalid_price == 0
    ])

    return (passed / 4) * 100

score1 = calculate_quality_score(df1)
score2 = calculate_quality_score(df2)

comparison = pd.DataFrame({
    "Dataset": [
        "sales_cleaned_day3.csv",
        "sample_sales_2.csv"
    ],
    "Quality Score (%)": [score1, score2]
})

comparison["Category"] = comparison["Quality Score (%)"].apply(
    lambda score: "Excellent" if score >= 90
    else "Good" if score >= 70
    else "Average" if score >= 50
    else "Poor"
)

comparison.to_csv(
    "day22_multi_dataset_score_comparison.csv",
    index=False
)

print("Day 22 report recreated successfully!")
display(comparison)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.bar(
    comparison["Dataset"],
    comparison["Quality Score (%)"]
)

plt.title("Data Quality Score Comparison")
plt.xlabel("Dataset")
plt.ylabel("Quality Score (%)")
plt.ylim(0, 100)
plt.xticks(rotation=15, ha="right")

plt.tight_layout()
plt.show()
```
