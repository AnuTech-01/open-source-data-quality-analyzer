```python
import pandas as pd
import matplotlib.pyplot as plt
import os

print("Day 20: Quality Summary Dashboard")
print("Current folder:", os.getcwd())
```

```python
report = pd.read_csv("day19_automatic_quality_report.csv")

print("Report loaded successfully!")
print("Total rows:", len(report))

display(report.head(10))
```

```python
print("Quality Check Status Summary")

status_summary = report["Status"].value_counts()

display(status_summary)

print("\nTotal PASS checks:", (report["Status"] == "PASS").sum())
print("Total FAIL checks:", (report["Status"] == "FAIL").sum())
print("Total INFO checks:", (report["Status"] == "INFO").sum())
```

```python
status_summary = report["Status"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(status_summary.index, status_summary.values)

plt.title("Data Quality Check Status Summary")
plt.xlabel("Quality Check Status")
plt.ylabel("Number of Checks")

plt.tight_layout()
plt.show()
```

```python
status_summary = report["Status"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(status_summary.index, status_summary.values)

plt.title("Data Quality Check Status Summary")
plt.xlabel("Quality Check Status")
plt.ylabel("Number of Checks")

plt.tight_layout()

plt.savefig(
    "day20_quality_status_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Chart saved successfully!")
print("File: day20_quality_status_chart.png")
```
