```python
import pandas as pd

print("Day 14: Data Quality Score")
```

```python
df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset loaded successfully!")
print("Total records:", len(df))
print(df.head())
```

```python
missing = int(df.isnull().sum().sum())
duplicates = int(df.duplicated().sum())
invalid_quantity = int((df["quantity"] <= 0).sum())
invalid_price = int((df["price"] <= 0).sum())

total_checks = 4

passed_checks = sum([
    missing == 0,
    duplicates == 0,
    invalid_quantity == 0,
    invalid_price == 0
])

quality_score = (passed_checks / total_checks) * 100

print("DATA QUALITY SCORE")
print("------------------")
print("Missing Values:", missing)
print("Duplicate Rows:", duplicates)
print("Invalid Quantities:", invalid_quantity)
print("Invalid Prices:", invalid_price)
print("Passed Checks:", passed_checks, "/", total_checks)
print("Quality Score:", quality_score, "%")
```

```python
score_report = pd.DataFrame({
    "Metric": [
        "Missing Values",
        "Duplicate Rows",
        "Invalid Quantities",
        "Invalid Prices",
        "Passed Checks",
        "Total Checks",
        "Quality Score (%)"
    ],
    "Result": [
        missing,
        duplicates,
        invalid_quantity,
        invalid_price,
        passed_checks,
        total_checks,
        quality_score
    ]
})

score_report.to_csv("day14_quality_score_report.csv", index=False)

print("Quality score report saved successfully!")
print(score_report)
```

```python
import os
import pandas as pd

file_name = "day14_quality_score_report.csv"

if os.path.exists(file_name):
    saved_score = pd.read_csv(file_name)

    print("SUCCESS: Quality Score Report exists!")
    print(saved_score)

    score = saved_score.loc[
        saved_score["Metric"] == "Quality Score (%)",
        "Result"
    ].iloc[0]

    print(f"\nFinal Data Quality Score: {score}%")
else:
    print("Report not found. Please run Step 4 again.")
```
