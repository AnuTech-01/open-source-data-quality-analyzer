```python
import pandas as pd
import os

print("Day 21: Overall Data Quality Score")
print("Project folder:", os.getcwd())
```

```python

df = pd.read_csv("sales_cleaned_day3.csv")

print("Dataset loaded successfully!")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))

display(df)
```

```python

missing_values = int(df.isnull().sum().sum())
duplicate_rows = int(df.duplicated().sum())
invalid_quantities = int((df["quantity"] <= 0).sum())
invalid_prices = int((df["price"] <= 0).sum())

total_checks = 4

passed_checks = sum([
    missing_values == 0,
    duplicate_rows == 0,
    invalid_quantities == 0,
    invalid_prices == 0
])

quality_score = (passed_checks / total_checks) * 100

print("Missing Values:", missing_values)
print("Duplicate Rows:", duplicate_rows)
print("Invalid Quantities:", invalid_quantities)
print("Invalid Prices:", invalid_prices)
print("------------------------------")
print("Overall Quality Score:", quality_score, "%")
```

```python
if quality_score >= 90:
    category = "Excellent"
elif quality_score >= 70:
    category = "Good"
elif quality_score >= 50:
    category = "Average"
else:
    category = "Poor"

print("Overall Quality Score:", quality_score, "%")
print("Quality Category:", category)
```

```python
score_report = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Missing Values",
        "Duplicate Rows",
        "Invalid Quantities",
        "Invalid Prices",
        "Overall Quality Score (%)",
        "Quality Category"
    ],
    "Result": [
        len(df),
        missing_values,
        duplicate_rows,
        invalid_quantities,
        invalid_prices,
        quality_score,
        category
    ]
})

display(score_report)

score_report.to_csv(
    "day21_overall_quality_score_report.csv",
    index=False
)

print("Report saved successfully!")
```
