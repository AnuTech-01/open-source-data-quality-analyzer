```python
import pandas as pd
import os
import glob

from data_quality_analyzer import check_data_quality

print("Day 19: Automatic Data Quality Analyzer")
print("Current folder:", os.getcwd())
```

```python

csv_files = glob.glob("*.csv")

print("Total CSV files found:", len(csv_files))
print("\nCSV Files List:")

for file in csv_files:
    print(file)
```

```python

all_reports = []

for file in csv_files:
    try:
        df = pd.read_csv(file)

        if "quantity" in df.columns and "price" in df.columns:
            report = check_data_quality(df)
            report["Dataset"] = file
            all_reports.append(report)

            print(f"Analyzed successfully: {file}")
        else:
            print(f"Skipped (required columns missing): {file}")

    except Exception as e:
        print(f"Could not analyze {file}: {e}")

print("\nAutomatic analysis completed!")
print("Datasets analyzed:", len(all_reports))
```

```python

if all_reports:
    final_report = pd.concat(all_reports, ignore_index=True)

    print("All quality reports combined successfully!")
    display(final_report.head(10))

else:
    print("No datasets were analyzed.")
```

```python

if all_reports:
    final_report.to_csv(
        "day19_automatic_quality_report.csv",
        index=False
    )

    print("Report saved successfully!")
    print("File: day19_automatic_quality_report.csv")
else:
    print("No report available to save.")
```

```python
file_name = "day19_automatic_quality_report.csv"

if os.path.exists(file_name):
    saved_report = pd.read_csv(file_name)

    print("Day 19 Completed Successfully!")
    print("Report file:", file_name)
    print("Total report rows:", len(saved_report))
    print("Datasets analyzed:", saved_report["Dataset"].nunique())
else:
    print("Report file not found. Please run the save cell again.")
```
