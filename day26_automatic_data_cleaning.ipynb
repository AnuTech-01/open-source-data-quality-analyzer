```python
import pandas as pd
import os

print("Day 26: Automatic Data Cleaning")
print("Project folder:", os.getcwd())
```

```python
df = pd.read_csv("sample_sales_2.csv")

print("Dataset loaded successfully!")
print("Rows before cleaning:", len(df))
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

display(df)
```

```python
# Remove duplicate rows
df_cleaned = df.drop_duplicates().copy()

# Fill missing product and city values
df_cleaned["product"] = df_cleaned["product"].fillna("Unknown")
df_cleaned["city"] = df_cleaned["city"].fillna("Unknown")

# Fill missing quantity with median
df_cleaned["quantity"] = df_cleaned["quantity"].fillna(
    df_cleaned["quantity"].median()
)

# Remove records with zero or negative quantity/price
df_cleaned = df_cleaned[
    (df_cleaned["quantity"] > 0) &
    (df_cleaned["price"] > 0)
].copy()

print("Automatic cleaning completed!")
print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(df_cleaned))

display(df_cleaned)
```

```python
df_cleaned.to_csv("day26_cleaned_sales.csv", index=False)

print("Cleaned dataset saved successfully!")
print("File name: day26_cleaned_sales.csv")
print("Total records saved:", len(df_cleaned))
```

```python
cleaned_df = pd.read_csv("day26_cleaned_sales.csv")

print("Cleaned file loaded successfully!")
print("Total records:", len(cleaned_df))
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Invalid quantities:", (cleaned_df["quantity"] <= 0).sum())
print("Invalid prices:", (cleaned_df["price"] <= 0).sum())

display(cleaned_df)
```

```python
cleaning_report = pd.DataFrame({
    "Metric": [
        "Rows Before Cleaning",
        "Rows After Cleaning",
        "Rows Removed",
        "Missing Values Remaining",
        "Duplicate Rows Remaining",
        "Invalid Quantities Remaining",
        "Invalid Prices Remaining"
    ],
    "Result": [
        len(df),
        len(cleaned_df),
        len(df) - len(cleaned_df),
        int(cleaned_df.isnull().sum().sum()),
        int(cleaned_df.duplicated().sum()),
        int((cleaned_df["quantity"] <= 0).sum()),
        int((cleaned_df["price"] <= 0).sum())
    ]
})

display(cleaning_report)
```
