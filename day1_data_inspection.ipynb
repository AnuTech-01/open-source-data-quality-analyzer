```python

import pandas as pd

data = {
    "order_id": [101, 102, 103, 104, 104, 106],
    "product": ["Laptop", "Mouse", "Keyboard", None, None, "Monitor"],
    "quantity": [1, 2, 1, 3, 3, None],
    "price": [50000, 500, 1500, 2000, 2000, 8000],
    "city": ["Jaipur", "Delhi", "Jaipur", "Mumbai", "Mumbai", None]
}

df = pd.DataFrame(data)

df.to_csv("sample_sales.csv", index=False)

print("CSV file created successfully!")
```
