```python
import pandas as pd

def check_data_quality(df):

    missing = df.isnull().sum().sum()
    duplicates = df.duplicated().sum()
    invalid_quantity = (df["quantity"] <= 0).sum()
    invalid_price = (df["price"] <= 0).sum()

    report = pd.DataFrame({
        "Check": [
            "Total Records",
            "Missing Values",
            "Duplicate Rows",
            "Invalid Quantities",
            "Invalid Prices"
        ],
        "Result": [
            len(df),
            missing,
            duplicates,
            invalid_quantity,
            invalid_price
        ],
        "Status": [
            "INFO",
            "PASS" if missing == 0 else "FAIL",
            "PASS" if duplicates == 0 else "FAIL",
            "PASS" if invalid_quantity == 0 else "FAIL",
            "PASS" if invalid_price == 0 else "FAIL"
        ]
    })

    return report

print("Function defined successfully!")
```

```python
df2 = pd.read_csv("sample_sales_2.csv")

report2 = check_data_quality(df2)

print(report2.to_string(index=False))
```

```python
import pandas as pd

files = [
    "sales_cleaned_day3.csv",
    "sample_sales_2.csv"
]

for file in files:
    print(f"\nChecking file: {file}")

    dataset = pd.read_csv(file)
    report = check_data_quality(dataset)

    print(report.to_string(index=False))
```

```python
import pandas as pd

files = [
    "sales_cleaned_day3.csv",
    "sample_sales_2.csv"
]

all_reports = []

for file in files:
    dataset = pd.read_csv(file)

    report = check_data_quality(dataset)

    # Add the filename to the report
    report.insert(0, "File Name", file)

    all_reports.append(report)

# Combine both reports
combined_report = pd.concat(all_reports, ignore_index=True)

print(combined_report.to_string(index=False))
```

```python
combined_report.to_csv(
    "day8_combined_quality_report.csv",
    index=False
)

print("Combined report saved successfully!")
```
