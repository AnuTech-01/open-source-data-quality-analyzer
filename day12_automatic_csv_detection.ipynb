```python
import os
import glob

print("Day 12: Automatic CSV File Detection")
print("Current folder:", os.getcwd())
```

```python
csv_files = glob.glob("*.csv")

print("CSV files found:", len(csv_files))
print("\nList of CSV files:")

for file in csv_files:
    print(file)
    
```

```python
from data_quality_analyzer import check_data_quality
import pandas as pd

all_reports = []

for file in csv_files:
    try:
        df = pd.read_csv(file)

        # Skip files that don't have the required columns
        required_columns = {"quantity", "price"}

        if not required_columns.issubset(df.columns):
            print(f"Skipped: {file} (required columns missing)")
            continue

        report = check_data_quality(df)
        report.insert(0, "File Name", file)
        all_reports.append(report)

        print(f"Checked successfully: {file}")

    except Exception as e:
        print(f"Error in {file}: {e}")

if all_reports:
    automatic_report = pd.concat(all_reports, ignore_index=True)
    print("\nAUTOMATIC DATA QUALITY REPORT")
    print(automatic_report.to_string(index=False))
else:
    print("No suitable CSV files found.")
```

```python
if all_reports:
    automatic_report.to_csv(
        "day12_automatic_quality_report.csv",
        index=False
    )

    print("Automatic report saved successfully!")
    print("Filename: day12_automatic_quality_report.csv")
else:
    print("No report available to save.")
```
