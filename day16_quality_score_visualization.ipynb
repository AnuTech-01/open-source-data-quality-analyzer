```python
import pandas as pd
import matplotlib.pyplot as plt

print("Day 16: Quality Score Visualization")
```

```python
comparison = pd.read_csv("day15_quality_score_comparison.csv")

print("Comparison report loaded successfully!")
print(comparison)
```

```python

plt.figure(figsize=(8, 5))

plt.bar(
    comparison["File Name"],
    comparison["Quality Score (%)"]
)

plt.title("Data Quality Score Comparison")
plt.xlabel("Dataset")
plt.ylabel("Quality Score (%)")
plt.ylim(0, 100)
plt.xticks(rotation=15, ha="right")

plt.tight_layout()
plt.show()
```
