```python
import pandas as pd
import os

print("Day 30: Complete Data Quality Analyzer")
print("Project folder:", os.getcwd())
```

```python
def run_data_quality_analyzer(file_name):
    try:
        df = pd.read_csv(file_name)
        print("✅ File loaded:", file_name)

        required_columns = [
            "order_id", "product", "quantity", "price", "city"
        ]

        missing_columns = [
            col for col in required_columns
            if col not in df.columns
        ]

        if missing_columns:
            print("❌ Missing required columns:", missing_columns)
            return None

        report = pd.DataFrame({
            "Check": [
                "Total Records",
                "Missing Values",
                "Duplicate Rows",
                "Invalid Quantities",
                "Invalid Prices"
            ],
            "Result": [
                len(df),
                int(df.isnull().sum().sum()),
                int(df.duplicated().sum()),
                int((pd.to_numeric(df["quantity"], errors="coerce") <= 0).sum()),
                int((pd.to_numeric(df["price"], errors="coerce") <= 0).sum())
            ]
        })

        report["Status"] = report.apply(
            lambda row: (
                "INFO" if row["Check"] == "Total Records"
                else "PASS" if row["Result"] == 0
                else "FAIL"
            ),
            axis=1
        )

        return report

    except FileNotFoundError:
        print("❌ File not found:", file_name)
    except pd.errors.ParserError:
        print("❌ CSV format error.")
    except Exception as e:
        print("❌ Unexpected error:", e)

    return None

print("Complete analyzer function created successfully!")
```

```python
report = run_data_quality_analyzer("day26_cleaned_sales.csv")

if report is not None:
    print("\nData Quality Analysis Completed!")
    display(report)
else:
    print("Analysis could not be completed. Check the error above.")
```

```python
def save_quality_report(report, output_file):
    if report is None:
        print("❌ No report available to save.")
        return

    try:
        report.to_csv(output_file, index=False)
        print("✅ Report saved successfully!")
        print("File:", output_file)

    except Exception as e:
        print("❌ Could not save report:", e)


save_quality_report(
    report,
    "day30_final_quality_report.csv"
)
```
